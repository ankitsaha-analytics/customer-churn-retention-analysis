## Business Insights & Recommendations

### Insight 1 — Contract Type


**Finding:** Month-to-month customers have the highest churn rate at 42.71%, compared with 11.27% for one-year contracts and only 2.83% for two-year contracts.

**Business Interpretation:** Customers on longer-term contracts are associated with substantially lower churn, while month-to-month customers represent the most vulnerable contract segment.

**Recommendation:** The company should target month-to-month customers with incentives such as discounted annual plans, loyalty benefits, or upgrade offers to encourage migration toward longer-term contracts.


### Insight 2 — Customer Tenure

**Finding:** Customers in their first 12 months have the highest churn rate at 47.44%. Churn falls to 28.71% for customers with 13–24 months of tenure, 20.39% for 25–48 months, and 9.51% for 49–72 months.

**Business Interpretation:** Churn is concentrated among relatively new customers, while longer-tenure customers show substantially higher retention. This suggests that the early stage of the customer relationship is a critical retention period.

**Recommendation:** The company should strengthen its first-year retention strategy through improved onboarding, proactive support, early engagement campaigns, and targeted retention offers for customers showing signs of disengagement.


### Insight 3 — Internet Service

**Finding:** Fiber optic customers have the highest churn rate at 41.89%, compared with 18.96% for DSL customers and 7.40% for customers without internet service.

**Business Interpretation:** Fiber optic customers represent a significantly higher-risk internet service segment. The data indicates an association between fiber optic service and higher churn, although it does not identify the underlying cause.

**Recommendation:** The company should investigate the fiber optic customer experience, including pricing, service quality, reliability, support interactions, and competitor offerings. Retention campaigns could then be targeted toward fiber optic customers based on the issues identified.

### Insight 4 — Payment Method

**Finding:** Customers using electronic check have the highest churn rate at 45.29%, compared with 19.11% for mailed check, 16.71% for automatic bank transfer, and 15.24% for automatic credit card payments.

**Business Interpretation:** Electronic check users represent a substantially higher-churn payment segment, while customers using automatic payment methods show much lower churn rates.

**Recommendation:** The company should investigate why electronic check customers have higher churn and consider encouraging eligible customers to adopt automatic payment methods through simplified enrollment, communication of convenience benefits, or appropriate incentives.

### Insight 5 — Support & Security Services

**Finding:** Customers without Tech Support had a 41.64% churn rate compared with 15.17% among customers with Tech Support. Similarly, customers without Online Security had a 41.77% churn rate compared with 14.61% among customers with Online Security.

**Business Interpretation:** Customers who do not subscribe to Tech Support or Online Security show substantially higher churn, suggesting that adoption of these services is associated with stronger customer retention.

**Recommendation:** The company should identify eligible customers without Tech Support or Online Security and test targeted service bundles, introductory offers, and educational campaigns that communicate the value of these services. The company should also investigate whether customers without these services have other characteristics contributing to their higher churn.


### Insight 6 — Monthly Charges & High-Risk Customers

**Finding:** Churned customers had higher average Monthly Charges (74.44) than retained customers (61.27). A custom high-risk segment consisting of customers on month-to-month contracts, with tenure of 12 months or less and Monthly Charges above 80, contained 464 customers. Of these, 341 churned, resulting in a 73.49% churn rate.

**Business Interpretation:** Churn risk becomes particularly high when multiple risk characteristics occur together. Newer customers with month-to-month contracts and relatively high monthly charges represent a particularly vulnerable customer segment in this dataset.

**Recommendation:** The company should prioritize this segment for proactive retention efforts such as early customer check-ins, personalized offers, plan reviews, service bundles, and targeted contract-upgrade incentives. The effectiveness of these interventions should be measured through controlled retention campaigns.

## Executive Business Summary

The analysis identified an overall customer churn rate of approximately 26.54%, with 1,869 of 7,043 customers having churned.

Churn was particularly high among month-to-month customers (42.71%), customers within their first 12 months (47.68%), fiber optic users (41.89%), and customers paying by electronic check (45.29%).

Customers without Tech Support or Online Security also showed churn rates above 41%, while churned customers had higher average Monthly Charges (74.44) than retained customers (61.27).

A rule-based high-risk segment combining month-to-month contracts, tenure of 12 months or less, and Monthly Charges above 80 identified 464 customers, of whom 341 had churned — a churn rate of 73.49%.

These findings suggest that retention efforts should focus on early-tenure and month-to-month customers, particularly those with higher monthly charges. The company should also investigate the customer experience of fiber optic and electronic check users and test targeted retention initiatives such as onboarding improvements, service bundles, proactive support, and longer-term contract incentives.
